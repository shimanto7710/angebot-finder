# AngebotFinder DE — Fetch offers

`scripts/fetch_offers.py` contains the fetching, parsing, and quality checks. This notebook calls the same script functions.

Set `FETCH_DATA = True` below when you want to refresh the saved offers.


```python
import json
import sys
from pathlib import Path

PROJECT_ROOT = next(
    (folder for folder in (Path.cwd(), *Path.cwd().parents)
     if (folder / "scripts" / "fetch_offers.py").is_file()),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError("Open the notebook from inside the AngebotFinder DE project")

sys.path.insert(0, str(PROJECT_ROOT))
from scripts.fetch_offers import DEFAULT_DATA_PATH, refresh_all_offers_json

FETCH_DATA = False
```

## Refresh offers

```python
refresh_result = refresh_all_offers_json(enabled=FETCH_DATA)
print(json.dumps(refresh_result, ensure_ascii=False, indent=2))
```
